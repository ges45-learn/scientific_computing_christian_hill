```python
def str_vector(v): # defines a function that accepts a single input argument
 # type checks input, working only if a tuple or list
  assert type(v) is list or type(v) is tuple, \
  'argument to str_vector must be list or tuple'
  assert len(v) in (2,3), 'vector must be 2D or 3D in str vector' # checks dimensions
  unit_vectors = ['i', 'j', 'k'] # setup labels for cartersian axes
  s = [] # initiliase list, this will store the formatted strings for each component before they get joined
  for i, component in enumerate(v): # () #loops over each input vector v
    s.append('{}{}'.format(component, unit_vectors[i])) #formats the current vector component with its corresponding unit vector and adds it to s
  return ' + '.join(s).replace('+-', '-')

v = [3,4,5]
vector = str_vector(v)
print(vector)

```

The Core Concept: *Only wrap the exact line that you want to catch errors for, or separate them so each operation gets its own error handling*

When you wrap code in a try block, you are telling the program: "If an error happens inside here, jump straight to the except block to handle it."

If you stuff multiple lines of code into that single try block, any line inside it can trigger the except block. If two different operations throw the exact same type of error (like IOError), your code won't be able to tell which line actually failed—so it treats both failures identically.

    fi = open(filename, 'r')   # Line 1: Tries to open the file

    # Error handling logic...

The Problem
If Line 2 fails, the program still jumps to the except IOError block and tells the user "File not found!" — which is completely misleading. You might try to recover from a missing file by prompting for a new filename, but that won't fix a corrupt disk or bad read stream.

```python
# eg4 astronomical data

# 1) import libraries / packages
import math # to access pi

# 2) create dictionary of tuples
# mass (kg) and radius (km) for some astronomical bodies
body = {'Sun': (1.988e30, 6.955e5), 'Mercury': (3.301e23, 2440.), 'Venus': (4.867e24, 6052), 'Earth': (5.97e24, 6371.), 'Mars': (6.417e23, 3390.), 'Jupiter': (1.89e27, 69911.), 'Saturn': (5.683e26, 58232.), 'Uranus': (8.681e25, 25362.), 'Neptune': (1.024e26, 24622.)}

planets = list(body.keys())
planets.remove('Sun')

def calc_density(m,r):
  """ returns the density of a sphere width mass m and radius r """
  return m / (4/3 * math.pi * r**3)

rho = {}

# iterates through each planet retrieving mass and radius from the list
for planet in planets:
  m, r = body[planet]
  # calculate the density in g / cm3
  rho[planet] = calc_density(m*1000, r*1e5) # unit conversion

for planet, density in sorted (rho.items()):
  print('the density of {0} is {1:3.2f} g/cm3'.format(planet, density))

```

```python
import math as m

def primes(n):
    """
    Returns a list of all prime numbers less than or equal to n.
    Uses an odd-only Sieve of Eratosthenes to save 50% memory and time.
    """
    # Create a boolean array where sieve[i] represents odd number (2i + 1).
    # Since we ignore even numbers, length is n // 2.
    sieve = [True] * (n // 2)

    # Check odd numbers starting from 3 up to sqrt(n)
    for i in range(3, int(m.sqrt(n)) + 1, 2):
        # Index corresponding to odd number 'i' is i // 2
        if sieve[i // 2]:
            # First composite multiple to cross off is i * i
            k = i * i
            # Step by i (which skips even multiples, striding by 2*i in terms of odd numbers)
            sieve[k // 2 :: i] = [False] * ((n - k - 1) // (2 * i) + 1)

    # Reconstruct prime list: prepend 2, then map active indices back to odd numbers (2i + 1)
    return [2] + [2 * i + 1 for i in range(1, n // 2) if sieve[i]]

# --- Main Script Execution ---

n = 1000000

# Step 1: Generate all primes up to n as a set for O(1) lookup speed
P = set(primes(n)) # converted into set P

# Step 2: Generate candidate Mersenne numbers of the form (2^i - 1) <= n
A = []
for i in range(2, int(m.log(n + 1, 2) + 1)):
    A.append(2**i - 1)

# Step 3: Find Mersenne primes via set intersection (unindented out of the loop)
# we can take the intersection of this set with any iterable object using the intersection method, so there is no need to explicitly convert our second list of integers A into a set
M = P.intersection(A)

# Step 4: Output the result sorted in ascending order
print(sorted(list(M)))
# Output: [3, 7, 31, 127, 8191, 131071]
# M is unordered collection of unique items by definition of a set ==> converted into a sorted list
```

```python
# e.4.8
#Modify Example E4.8 to use a defaultdict to produce a list of words, keyed by their length from the text of the first line of the Gettysburg Address.

import collections # to access missing keys so we dont get KeyError

text = 'Four score and seven years ago our fathers brought forth on this continent, a new nation, conceived in Liberty, and dedicated to the proportion that all men are created equal'

# creates an empty dictionary-like object with int ensuring 0 is the default
#splits sentence into a list using spaces as the delimiter
# for iterates through
word_groups = collections.defaultdict(list)
for word in text.split():
  # len(word) calculates the number of characters in the current
  # uses the lenght as the dictionary key
  # += 1 increments the counter for that word length by 1
  word_groups[len(word)].append(word)

#outputs the final dicitonary displaying {word_length: total_count}

print(word_groups)

#
```

```python
import collections

text = 'Four score and seven years ago fathers brought forth this'\
'continent, a new nation, conceived in Liberty, and dedicated to'\
'the proposition that all men are created equal.'

word_lengths = collections.defaultdict(int)

for word in text.split():
  word_lengths[len(word)] += 1

for word, count in word_lengths.items():
  print('{}: {}'.format(word, count))
# output: 2 words have 4 letters
#
```
