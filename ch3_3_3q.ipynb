```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure('Q3_3')
ax = fig.add_subplot(111)


#parameterss
T1 = 200 # K
T2 = 300 # K
T3 = 400 # K
kb = 8.63e-5 # eV / K
x = np.linspace(-0.2,0.2,50)
y200 = (1+np.exp(x/(kb*T1)))**-1
y300 = (1+np.exp(x/(kb*T2)))**-1
y400 = (1+np.exp(x/(kb*T3)))**-1

ax.plot(x,y200,label='T=200K')
ax.plot(x,y300,label='T=300K')
ax.plot(x,y400,label='T=400K')
```
